````python
import os
import json
import pandas as pd

# --- Local (with a .env file) ---
from dotenv import load_dotenv
load_dotenv()
API_KEY = os.environ["GROQ_API_KEY"]



# TODO: set API_KEY using ONE of the methods above.

# OpenAI-compatible client (works for Groq and OpenAI; Gemini users see their docs):
from openai import OpenAI

client = OpenAI(
    api_key=API_KEY,
    base_url="https://api.groq.com/openai/v1",  
)
MODEL = "llama-3.3-70b-versatile"                # or your provider's model name

print("Client ready.")
````

````python
def ask_llm(user_prompt, system_prompt="You are a helpful assistant.",
            temperature=0.7, max_tokens=500):
     response = client.chat.completions.create(
         model=MODEL,
         messages=[
             {"role": "system", "content": system_prompt},
             {"role": "user",   "content": user_prompt},
         ],
         temperature=temperature,
         max_tokens=max_tokens,
     )
     return response

# TODO: Call it once with a simple question and print the answer.
response = ask_llm("What is family therapy?")
print(response.choices[0].message.content)

# TODO: Print response.usage as well — how many tokens did your call consume?
print(response.usage)

     
````

**Difference between system and user**

**system**:tells the AI its overall instructions, behavior, or rules for how it should respond.

**user**: Contains the actual question or request from the person using the AI

**What is Token**
A token is a small piece of text that an LLM processes. A token can be a whole word, part of a word, punctuation, or a space-related piece.

**Why do API providers bill per token rather than per request?**
API providers bill based on tokens because the amount of text the model processes varies greatly between requests. A short request might use only a few tokens, while a long conversation or document could use thousands or millions.

Therefore, charging per token is more accurate and fair than charging simply per request, because two requests can require vastly different amounts of computation.

````python
question = "what is dark matter?"

# Temperature = 0.0
print("===== Temperature 0.0 =====")

for i in range(5):
    answer = ask_llm(
        question,
        temperature=0.0
    )
    print(f"{i + 1}. {answer}")


# Temperature = 1.2
print("\n===== Temperature 1.2 =====")

for i in range(5):
    answer = ask_llm(
        question,
        temperature=1.2
    )
    print(f"{i + 1}. {answer}")
````

**At temperature 0.0**, the responses were generally very similar or sometimes identical. The model was more consistent and predictable.

**At temperature 1.2**, the responses were more varied and creative. The model produced different suggestions and wording across the five attempts.

**For the loan decision-support system** a low temperature, preferably around 0.0–0.3, is more appropriate. Loan decisions require consistency, predictability, and reliability rather than creative or random responses. Using a higher temperature could cause the system to give different recommendations for similar inputs, which would be undesirable for a financial decision-support system.

**Why is "no invented details" essential? What is this failure mode called?**
The instruction is essential because a loan decision-support system should base its analysis only on information actually provided by the applicant. If the LLM invents or assumes facts, it could incorrectly make an applicant appear more or less financially reliable and potentially influence a real lending decision.

This failure mode is commonly called an LLM hallucination (or hallucination). It occurs when a model generates information that is unsupported by, or not present in, the source material

````python

EXTRACT_PROMPT = """
Extract information from the loan application and return ONLY a valid JSON object.

The JSON object MUST contain EXACTLY these six keys:

{
  "applicant_name": "string",
  "amount_ghs": number,
  "purpose": "string",
  "monthly_profit_ghs": number or null,
  "has_collateral_or_guarantor": true or false,
  "repayment_months": number or null
}

Rules:
1. Return ONLY the JSON object.
2. Do not include explanations, comments, or markdown.
3. Use EXACTLY the six keys shown above.
4. If a field is not stated in the letter, use null.
5. Do not guess, assume, or invent any information.
6. amount_ghs must be a number, not a string.
7. monthly_profit_ghs must be a number or null.
8. repayment_months must be a number or null.
9. has_collateral_or_guarantor must be true or false.
10. If either collateral OR a guarantor is mentioned, set
    has_collateral_or_guarantor to true.
11. Keep the purpose concise but faithful to the application.

ONE WORKED EXAMPLE:

Letter:
My name is Ama Mensah. I operate a small bakery in Tema and have been
in business for four years. I am requesting GHS 10,000 to purchase a
new oven and additional baking supplies. My average monthly profit is
GHS 1,500. My brother has agreed to act as my guarantor. I propose to
repay the loan over 10 months.

Correct JSON:
{
  "applicant_name": "Ama Mensah",
  "amount_ghs": 10000,
  "purpose": "purchase a new oven and additional baking supplies",
  "monthly_profit_ghs": 1500,
  "has_collateral_or_guarantor": true,
  "repayment_months": 10
}

Now extract the fields from this loan application:

LETTER_TEXT_HERE
"""



# EXTRACTION FUNCTION


def extract_fields(letter_text):

    # Use replace() instead of .format()
    # This prevents JSON braces from causing KeyError
    prompt = EXTRACT_PROMPT.replace(
        "LETTER_TEXT_HERE",
        letter_text
    )

    try:

        response = ask_llm(
            prompt,
            temperature=0.0
        )

        # Get the model's actual text response
        result = response.choices[0].message.content.strip()

        # Print the raw response if needed for debugging
        # print("RAW RESPONSE:", result)

        
        # Remove ```json fences
        

        if result.startswith("```json"):
            result = result[7:].strip()

        elif result.startswith("```"):
            result = result[3:].strip()

        if result.endswith("```"):
            result = result[:-3].strip()

      
        # Parse JSON
       

        data = json.loads(result)

       
        # Make sure response is a dictionary
       

        if not isinstance(data, dict):
            print("WARNING: Response is not a JSON object.")
            return None

       
        # Required keys
       

        required_keys = {
            "applicant_name",
            "amount_ghs",
            "purpose",
            "monthly_profit_ghs",
            "has_collateral_or_guarantor",
            "repayment_months"
        }

        # Check that exactly the required keys exist
        if set(data.keys()) != required_keys:
            print("WARNING: JSON does not contain exactly the required keys.")
            print("Returned keys:", list(data.keys()))
            return None

        return data

    except json.JSONDecodeError:
        print("WARNING: Could not parse model response as JSON.")
        print("Model response:")
        print(result)
        return None

    except Exception as e:
        print(f"WARNING: Extraction failed: {e}")
        return None



# 3. TEST ONE LETTER FIRST


print("Testing extraction on L001...\n")

test_result = extract_fields(LETTERS["L001"])

print("Extracted result:")
print(test_result)



# RUN ON ALL SIX LETTERS


results = []

print("\n" + "=" * 70)
print("EXTRACTING ALL SIX LETTERS")
print("=" * 70)

for letter_id, letter_text in LETTERS.items():

    print(f"\nExtracting {letter_id}...")

    extracted = extract_fields(letter_text)

    if extracted is not None:

        extracted["letter_id"] = letter_id

        results.append(extracted)

        print("Extraction successful.")

    else:

        print("Extraction failed.")



# CREATE DATAFRAME


df_extracted = pd.DataFrame(results)



# ORDER COLUMNS


columns = [
    "letter_id",
    "applicant_name",
    "amount_ghs",
    "purpose",
    "monthly_profit_ghs",
    "has_collateral_or_guarantor",
    "repayment_months"
]

# Only reorder if we actually have results
if not df_extracted.empty:
    df_extracted = df_extracted[columns]

# DISPLAY RESULTS

print("\n" + "=" * 70)
print("FINAL EXTRACTED DATA")
print("=" * 70)

display(df_extracted)
````

**Why must the few-shot example NOT come from the six letters being processed?**
because it could leak the expected answers into the model's reasoning. If the model sees one of the actual test letters as a worked example, it may simply copy the structure or values from that letter instead of demonstrating that it can independently extract information.sing a separate example tests whether the prompt works generally, rather than only working because the model has already seen one of the answers.

**Why “use null, do not guess”? What did the model do without that instruction?**
because the model may try to fill in missing information based on assumptions. For example, L002 does not state Kwame's monthly profit or a specific repayment period. Without the instruction to use null, the model might try to infer or invent values.

**Why is temperature=0 appropriate for extraction but arguably not for creative tasks?**
because we want the model to produce consistent and predictable results. Given the same loan application, we want the same fields and values extracted each time. Creativity or variation is not useful when converting a letter into structured financial data.

````python

BRIEF_PROMPT = """
You are an assistant supporting a microfinance loan officer.

Prepare an evidence-based recommendation brief using ONLY:
1. The original loan application.
2. The extracted JSON information.

The purpose of this brief is to SUPPORT a human loan officer.
The AI MUST NOT make the final lending decision.

IMPORTANT RULES:
- Do not invent, assume, or infer information.
- Every strength and risk must be directly supported by the letter.
- Do not turn opinions or claims made by the applicant into verified facts.
- Do not treat age, enthusiasm, optimism, trustworthiness, or similar
  personal claims as evidence of repayment ability unless independently
  supported by information in the letter.
- Do not describe something as a strength merely because it exists.
- If a fact could reasonably be either positive or negative, describe it
  neutrally rather than labeling it a strength.
- Do not use information that is not in the application.
- If information is missing, explicitly identify it.
- Do not invent financial figures, business experience, collateral value,
  repayment ability, or documents.
- Do NOT recommend "approve" or "reject".
- The final decision MUST be made by a human loan officer.

Use EXACTLY these four sections:

## 1. Strengths
- List only concrete, evidence-based strengths.
- Each strength must be directly supported by the application.
- Do not infer future success from age, enthusiasm, optimism, or intentions.

## 2. Risks / Red Flags
- List concrete risks or concerns directly supported by the application.
- Focus on financial uncertainty, repayment concerns, lack of business history,
  lack of collateral, existing debts, or other relevant evidence.
- Do not exaggerate or invent risks.

## 3. Missing Information
- List information or documents that the loan officer should request
  before making a decision.
- Examples include business records, bank statements, proof of income,
  business registration, collateral documentation, guarantor information,
  or a business plan when relevant.
- Only request information that is relevant to the application.

## 4. Suggested Next Step
- Suggest ONE practical action for the human loan officer.
- Examples:
  "Invite the applicant for an interview."
  "Request supporting financial documents."
  "Verify the guarantor and collateral."
  "Request a detailed business plan."
  "Flag for senior review."
- NEVER say "approve" or "reject".

ORIGINAL LOAN APPLICATION:
{letter}

EXTRACTED INFORMATION:
{extracted_json}
"""



# FUNCTION TO GENERATE BRIEF


def generate_brief(letter_text, extracted_data):

    extracted_json = json.dumps(
        extracted_data,
        indent=2
    )

    # Use replace() so JSON braces do not cause formatting errors
    prompt = BRIEF_PROMPT.replace(
        "{letter}",
        letter_text
    ).replace(
        "{extracted_json}",
        extracted_json
    )

    try:

        response = ask_llm(
            prompt,
            temperature=0.0
        )

        return response.choices[0].message.content.strip()

    except Exception as e:

        print(f"WARNING: Could not generate brief: {e}")
        return None


# GENERATE BRIEFS FOR ALL SIX LETTERS


briefs = {}

for letter_id, letter_text in LETTERS.items():

    print(f"Generating brief for {letter_id}...")

    row = df_extracted[
        df_extracted["letter_id"] == letter_id
    ]

    if row.empty:
        print(f"WARNING: No extracted data found for {letter_id}")
        briefs[letter_id] = None
        continue

    extracted_data = row.iloc[0].to_dict()

    # Remove letter_id from extracted JSON
    extracted_data.pop("letter_id", None)

    briefs[letter_id] = generate_brief(
        letter_text,
        extracted_data
    )



# PRINT L001, L002 AND L006


for letter_id in ["L001", "L002", "L006"]:

    print("\n" + "=" * 80)
    print(f"LOAN BRIEF — {letter_id}")
    print("=" * 80)

    if briefs[letter_id]:
        print(briefs[letter_id])
    else:
        print("No brief generated.")
````

These are concrete pieces of evidence that support the application. L006 is much weaker based on the information provided.Kofi has not started any of the three proposed businesses, has no collateral, provides no existing business revenue or profit, and wants to repay the GHS 50,000 in one year when the businesses become successful. **Why forbid “approve” / “reject”?**

AI identifies evidence, strengths, risks, and missing information while the human loan officer makes the final decision.

````python

# GOLD-STANDARD EVALUATION

# The letters we want to evaluate
gold_letters = ["L001", "L003", "L006"]


# Fields to compare
fields = [
    "applicant_name",
    "amount_ghs",
    "purpose",
    "monthly_profit_ghs",
    "has_collateral_or_guarantor",
    "repayment_months"
]
# HELPER FUNCTION FOR COMPARING VALUES


def values_match(field, predicted, actual):

    # Name comparison: case-insensitive
    if field == "applicant_name":

        if predicted is None or actual is None:
            return predicted == actual

        return str(predicted).strip().lower() == str(actual).strip().lower()

    # Purpose comparison: case-insensitive
    elif field == "purpose":

        if predicted is None or actual is None:
            return predicted == actual

        return str(predicted).strip().lower() == str(actual).strip().lower()

    # Numbers and booleans must match exactly
    else:
        return predicted == actual

# BUILD COMPARISON TABLE

comparison_rows = []


for field in fields:

    row = {
        "field": field
    }

    correct_count = 0

    for letter_id in gold_letters:

        # Get the gold-standard value
        gold_value = GOLD[letter_id][field]

        # Get the extracted value from the DataFrame
        extracted_row = df_extracted[
            df_extracted["letter_id"] == letter_id
        ]

        if extracted_row.empty:

            predicted_value = None

        else:

            predicted_value = extracted_row.iloc[0][field]

        # Compare
        match = values_match(
            field,
            predicted_value,
            gold_value
        )

        # Store result as ✓ / ✗
        row[letter_id] = "✓" if match else "✗"

        if match:
            correct_count += 1

    # Calculate accuracy across the 3 letters
    row["accuracy"] = correct_count / len(gold_letters)

    comparison_rows.append(row)


# CREATE DATAFRAME

accuracy_df = pd.DataFrame(comparison_rows)


# FORMAT ACCURACY AS PERCENTAGE

accuracy_df["accuracy"] = (
    accuracy_df["accuracy"] * 100
).map(lambda x: f"{x:.1f}%")

# DISPLAY RESULT

print("=" * 70)
print("EXTRACTION ACCURACY AGAINST GOLD STANDARD")
print("=" * 70)

display(accuracy_df)
````

````python

# TEMPERATURE CONSISTENCY TEST


letter_id = "L004"
letter_text = LETTERS[letter_id]


# FUNCTION TO RUN EXTRACTION MULTIPLE TIMES


def run_consistency_test(temperature, runs=5):

    results = []
    valid_json_count = 0

    print("=" * 70)
    print(f"TEMPERATURE = {temperature}")
    print("=" * 70)

    for i in range(runs):

        print(f"\nRun {i + 1}:")

        # Create the prompt directly
        prompt = EXTRACT_PROMPT.replace(
            "LETTER_TEXT_HERE",
            letter_text
        )

        try:

            response = ask_llm(
                prompt,
                temperature=temperature
            )

            raw_result = response.choices[0].message.content.strip()

            # Remove markdown JSON fences
            if raw_result.startswith("```json"):
                raw_result = raw_result[7:].strip()

            elif raw_result.startswith("```"):
                raw_result = raw_result[3:].strip()

            if raw_result.endswith("```"):
                raw_result = raw_result[:-3].strip()

            # Try to parse JSON
            result = json.loads(raw_result)

            if isinstance(result, dict):

                valid_json_count += 1
                results.append(result)

                print("Valid JSON: YES")
                print(result)

            else:

                print("Valid JSON: NO")

        except Exception as e:

            print("Valid JSON: NO")
            print(f"Error: {e}")


    # COUNT UNIQUE RESULTS
    

    unique_results = set()

    for result in results:

        result_string = json.dumps(
            result,
            sort_keys=True
        )

        unique_results.add(result_string)


    # REPORT RESULTS

    print("\n" + "-" * 70)
    print(f"Temperature: {temperature}")
    print(f"Valid JSON runs: {valid_json_count}/{runs}")
    print(f"Unique valid outputs: {len(unique_results)}")

    if valid_json_count == runs and len(unique_results) == 1:
        print("Identical across all 5 runs: YES")

    else:
        print("Identical across all 5 runs: NO")

    return results, unique_results


# TEMPERATURE 0

results_temp_0, unique_temp_0 = run_consistency_test(
    temperature=0.0,
    runs=5
)

# TEMPERATURE 1.0


results_temp_1, unique_temp_1 = run_consistency_test(
    temperature=1.0,
    runs=5
)

# FINAL SUMMARY


print("\n\n")
print("=" * 70)
print("FINAL CONSISTENCY SUMMARY")
print("=" * 70)

print(
    f"Temperature 0.0: "
    f"{len(results_temp_0)}/5 valid JSON, "
    f"{len(unique_temp_0)} unique output(s)"
)

print(
    f"Temperature 1.0: "
    f"{len(results_temp_1)}/5 valid JSON, "
    f"{len(unique_temp_1)} unique output(s)"
)
````

**Audit logging and an appeal process**
